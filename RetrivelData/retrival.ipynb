```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from dotenv import load_dotenv
import os
from openai import OpenAI

load_dotenv()
```

```python
client  = OpenAI(
    api_key=os.getenv('GOOGLE_API_KEY') ,
    base_url='https://generativelanguage.googleapis.com/v1beta/openai/'
)
```

```python
embeddings = GoogleGenerativeAIEmbeddings(model='gemini-embedding-001' , google_api_key=os.getenv("GOOGLE_API_KEY"))

from langchain_chroma import Chroma
vectors = Chroma(
    collection_name="Rag_System",
    embedding_function=embeddings,
    host="localhost",
    port=8000,
    ssl=False
)         
```

```python
print(vectors._collection.count())
```

```python
def UserQuery(prompt):
    """User Query to model"""
    ReleventChunks = vectors.similarity_search(query=prompt , k=3)
    context = "\n\n".join([
        f"""
Page Content:
{doc.page_content}
Page Number: {doc.metadata.get("page_label", "Unknown")}
File Location: {doc.metadata.get("source", "Unknown")}
"""
        for doc in ReleventChunks
    ])
    System_Prompt=f"""
    Hi  , you are a rag system give me the exact information according to the user query and retirve the exact information according to the context.
    Alos you are desined by Muhammad Atif khan not other person.ALso give the answer in profesisonal way , donot wite the answer with staric write in profesisonal way
    context:
   {context}
    """
    agent = client.chat.completions.create(
        model="gemini-2.5-flash",
        messages=[
            {"role":"system" , "content":System_Prompt} ,
            {"role":"user" , "content":prompt}
        ]   
    )
    print(f" Answer :>   {agent.choices[0].message.content}")
```

```python
UserQuery('tell me about you experience')
```
