```python
import os
from dotenv import load_dotenv
from langchain_community.document_loaders import PyMuPDFLoader , PyPDFLoader , DirectoryLoader
from pathlib import Path
load_dotenv()
print(os.getenv("GOOGLE_API_KEY"))
```

```python
# make a function which take pdfs and convert it into a directory

def Process_all_pdf_files(pdf_files):
    """process all the pdf files into a dictionary"""
    all_docx =[]
    pdf_dir = Path(pdf_files)

    load = list(pdf_dir.glob('**/*.pdf'))

    print(f"founded pdf files {len(load)} to process")

    for pdf_fle in load:
        print(f"name of all pdf files {pdf_fle.name}")
        try:
         loader = PyPDFLoader(str(pdf_fle))
         docx  = loader.load()

         # also added meta data 
         for doc in docx:
            doc.metadata['source_data'] = pdf_fle.name
            doc.metadata['file_type'] = 'pdf'

         all_docx.extend(docx)
         print(f"leangth of document {len(docx)}")

        except Exception as e:
            print(f"Error => {e}")
    return all_docx       
all_document = Process_all_pdf_files('../Data')
```

```python
from h11._abnf import chunk_size
from langchain_text_splitters import RecursiveCharacterTextSplitter

def Chunking(documents , chunk_sze = 800 , chunk_over=150):

 """COnverting the documents into chunks for better rag performance"""
 text_splitter = RecursiveCharacterTextSplitter(
 chunk_size = chunk_sze ,
 chunk_overlap = chunk_over,
 length_function=len ,
 separators=["/n/n" , "/n" , " " , ""]

    )

 Text_to_chunks= text_splitter.split_documents(documents)
 print(f"split document {len(documents)} into chunks {len(Text_to_chunks)}")

 if Text_to_chunks:
    print(f"/n Example of chunks")
    print(f"content : {Text_to_chunks[0].page_content[:200]} ")
    print(f"MetaData : {Text_to_chunks[0].metadata}")
 return Text_to_chunks
```

```python
chunk = Chunking(all_document)
chunk
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings

embeddings = GoogleGenerativeAIEmbeddings(model='gemini-embedding-001' , google_api_key=os.getenv("GOOGLE_API_KEY"))
```

```python
from langchain_chroma import Chroma

vectors = Chroma(
    collection_name='Rag_System' ,
    embedding_function=embeddings ,
    persist_directory="./chroma_db"
)

ids = vectors.add_documents(chunk)
print(f"Stored {len(ids)} chunks")
```
