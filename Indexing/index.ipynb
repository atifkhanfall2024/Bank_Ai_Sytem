# RAG Indexing Notebook (Google Gemini)
This notebook loads documents from the `Data` directory and indexes them using LangChain and Google Gemini API.

```python
from dotenv import load_dotenv
import os

# Load environment variables from .env file
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model

agent = init_chat_model(
    model='llama-3.3-70b-versatile',
    model_provider='groq'
)
```

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader('../Data/Muhammad_Atif_Khan.txt' , encoding='utf-8')
document = loader.load()
```

```python
from langchain_community.document_loaders import DirectoryLoader , PyMuPDFLoader , PyPDFLoader

dir_load = DirectoryLoader(
    '../Data' ,  glob='**/* .pdf',
    loader_cls=PyMuPDFLoader ,
    loader_kwargs={'encoding':"utf-8"} ,
    show_progress=True
)

docx = dir_load.load()
print(f'loaded page :  {len(docx)}')
```

```python

from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 800 ,
    chunk_overlap = 150 ,
)

text_into_chunks = text_splitter.split_documents(documents=document)
```
